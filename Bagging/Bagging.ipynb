```python
import pandas as pd
from sklearn.model_selection import train_test_split as ttp
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
df.nunique()
```

```python
df.info()
```

## Removing outliers

```python
restingbp_mean = df.RestingBP.mean()
restingbp_std = df.RestingBP.std()
restingbp_upper_limit = restingbp_mean + 3*restingbp_std
restingbp_lower_limit = restingbp_mean - 3*restingbp_std

df1 = df[(df.RestingBP < restingbp_upper_limit) & (df.RestingBP > restingbp_lower_limit)]
df1
```

```python
cholesterol_mean = df.Cholesterol.mean()
cholesterol_std = df.Cholesterol.std()
cholesterol_upper_limit = cholesterol_mean + 3*cholesterol_std
cholesterol_lower_limit = cholesterol_mean - 3*cholesterol_std

df2 = df1[(df.Cholesterol < cholesterol_upper_limit) & (df.Cholesterol > cholesterol_lower_limit)]
df2
```

```python
oldpeak_mean = df.Oldpeak.mean()
oldpeak_std = df.Oldpeak.std()
oldpeak_upper_limit = oldpeak_mean + 3*oldpeak_std
oldpeak_lower_limit = oldpeak_mean - 3*oldpeak_std

df3 = df2[(df.Oldpeak < oldpeak_upper_limit) & (df.Oldpeak > oldpeak_lower_limit)]
df3
```

```python
df3.shape
```

```python
dataset = pd.get_dummies(df3, drop_first = True)
dataset
```

```python
X = dataset.drop('HeartDisease', axis = 1)
y = dataset.HeartDisease

X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
X_train, X_test, y_train, y_test = ttp(X_scaled,y, test_size = 0.2, random_state = 42)
X_train.shape 
```

```python
X_test.shape
```

```python
svm_model = SVC()
dtc_model = DecisionTreeClassifier()
```

# SVM classifier stand-alone and bag model

```python
svm_model.fit(X_train,y_train)
svm_model.score(X_test,y_test)
```

```python
bag_model_svm = BaggingClassifier(
    estimator = svm_model,
    n_estimators = 100,
    max_samples = 0.8,
    oob_score = True,
    random_state = 42
)

bag_model_svm.fit(X_train,y_train)
bag_model_svm.oob_score_
```

# Decision tree classifier stand-alone and bag model

```python
dtc_model.fit(X_train,y_train)
dtc_model.score(X_test,y_test)
```

```python
bag_model_dtc = BaggingClassifier(
    estimator = dtc_model,
    n_estimators = 100,
    max_samples = 0.8,
    oob_score = True,
    random_state = 42
)

bag_model_dtc.fit(X_train,y_train)
bag_model_dtc.oob_score_
```

# Bagging has been used to improve the performance of the decision tree algorithm. It is a technique that involves training multiple decision trees on different subsets of the training data and then combining their predictions to produce a final prediction. The idea behind bagging is that by training multiple trees on different subsets of the data, the trees will be less prone to overfitting and will be more robust to noise in the data. This can lead to improved performance on unseen data.

```python
from sklearn.model_selection import cross_val_score
```

```python
bag_model = BaggingClassifier(
    estimator = dtc_model,
    n_estimators = 100,
    max_samples = 0.8,
    oob_score = True,
    random_state = 42
)

scores = cross_val_score(bag_model, X,y, cv = 5)
scores
```
