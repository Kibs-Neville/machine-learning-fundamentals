```python
import pandas as pd

df = pd.read_csv("AB_NYC_2019.csv")
df.head()
```

```python
max_threshold = df.price.quantile(0.95)
min_threshold = df.price.quantile(0.05)

print(max_threshold)
print(min_threshold)
```

```python
df[df.price < min_threshold]
```

```python
df[df.price > max_threshold]
```

```python
df_with_outliers = df[(df.price < min_threshold) | (df.price > max_threshold)]
df_with_outliers
```

```python
df_no_outliers = df[(df.price < max_threshold) & (df.price > min_threshold)]
df_no_outliers
```
